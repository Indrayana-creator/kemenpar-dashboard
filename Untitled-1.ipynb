```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(
    r"C:/Users/use/Downloads/master_demand_supply_merged.csv"
)

print(df.shape)
```

```python
# Simpan nama destinasi
destinasi = df["DPP"]

# Hapus kolom identitas
X = df.drop(columns=["DPP"])

# Isi missing value dengan median
imputer = SimpleImputer(strategy="median")
X = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns
)

# Standardisasi
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

hasil = []

for k in range(2, len(df)):

    # K-Means
    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    km_labels = km.fit_predict(X_scaled)

    km_score = silhouette_score(
        X_scaled,
        km_labels
    )

    # Hierarchical
    hc = AgglomerativeClustering(
        n_clusters=k,
        linkage="ward"
    )

    hc_labels = hc.fit_predict(X_scaled)

    hc_score = silhouette_score(
        X_scaled,
        hc_labels
    )

    hasil.append([
        k,
        km_score,
        hc_score
    ])

hasil_df = pd.DataFrame(
    hasil,
    columns=[
        "K",
        "KMeans",
        "Hierarchical"
    ]
)

print(hasil_df)
```

Pada kedua metode, nilai Silhouette Score tertinggi diperoleh pada K=3 sebesar 0,184447. Ketika jumlah cluster ditingkatkan menjadi K=4 hingga K=9, nilai Silhouette Score cenderung menurun. Hal ini menunjukkan bahwa pembagian data menjadi tiga kelompok menghasilkan pemisahan cluster yang lebih baik dibandingkan jumlah cluster lainnya.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(
    hasil_df["K"],
    hasil_df["KMeans"],
    marker="o",
    label="K-Means"
)

plt.plot(
    hasil_df["K"],
    hasil_df["Hierarchical"],
    marker="s",
    label="Hierarchical"
)

plt.xlabel("Jumlah Cluster (K)")
plt.ylabel("Silhouette Score")
plt.title("Perbandingan Silhouette Score")
plt.legend()
plt.grid(True)

plt.show()
```

```python
print("\n=== KMEANS TERBAIK ===")
print(
    hasil_df.loc[
        hasil_df["KMeans"].idxmax()
    ]
)

print("\n=== HIERARCHICAL TERBAIK ===")
print(
    hasil_df.loc[
        hasil_df["Hierarchical"].idxmax()
    ]
)
```

Kenapa pakai K=3

"Karena berdasarkan evaluasi Silhouette Score pada K=2 sampai K=9, nilai tertinggi diperoleh pada K=3 sebesar 0,184447 sehingga dipilih sebagai jumlah cluster optimal."

Kalau sudah sampai tahap: K terbaik = 3

maka jangan cari K lagi. Langkah berikutnya adalah membangun model final dengan K=3, lalu menginterpretasikan cluster.

```python
#Model Final Hierarchical Clustering (K = 3)

import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering


df = pd.read_csv(
    r"C:/Users/use/Downloads/master_demand_supply_merged.csv"
)


# PREPROCESSING

X = df.drop(columns=["DPP"])

# Mengatasi missing value
imputer = SimpleImputer(strategy="median")
X = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns
)

# Standardisasi
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# MODEL FINAL

final_model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

df["Cluster"] = final_model.fit_predict(X_scaled)

# HASIL CLUSTER

print(df[["DPP", "Cluster"]].sort_values("Cluster"))
```

```python
cluster_summary = df.groupby("Cluster").mean(
    numeric_only=True
)

print(cluster_summary)
```

```python
df.to_csv(
    r"C:/Users/use/Downloads/hasil_cluster_final.csv",
    index=False
)

print("Berhasil disimpan")
```

Berdasarkan perbandingan metode K-Means dan Hierarchical Clustering menggunakan Silhouette Score pada K=2 hingga K=9, nilai tertinggi diperoleh pada K=3 sebesar 0,184447. Karena jumlah destinasi yang dianalisis relatif sedikit (10 destinasi), Hierarchical Clustering dipilih sebagai metode utama. Selanjutnya dilakukan pemodelan akhir menggunakan Hierarchical Clustering dengan jumlah cluster sebanyak 3 untuk mengelompokkan destinasi wisata berdasarkan karakteristik supply dan demand.

Hasil clustering menunjukkan bahwa destinasi wisata dapat dikelompokkan menjadi tiga cluster. Cluster pertama terdiri atas Borobudur, Likupang, Mandalika, Morotai, dan Wakatobi yang memiliki aksesibilitas dan kualitas akomodasi yang relatif baik. Cluster kedua terdiri atas Bromo, Danau Toba, Labuan Bajo, dan Raja Ampat yang memiliki tingkat popularitas serta dukungan fasilitas wisata yang lebih tinggi. Sementara itu, cluster ketiga hanya terdiri atas Tanjung Kelayang yang memiliki tingkat ketertarikan wisatawan yang tinggi namun masih menghadapi keterbatasan aksesibilitas. Berdasarkan hasil tersebut, setiap cluster menunjukkan karakteristik yang berbeda sehingga memerlukan strategi pengembangan pariwisata yang berbeda pula.
